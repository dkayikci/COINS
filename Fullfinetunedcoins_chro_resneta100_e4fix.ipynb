```python
# =====================================================================================
# CARIAN POTTERY — COMPLETE PIPELINE (single cell)
#
#   A. Classification with 4-fold nested cross-validation   -> the paper's metrics
#   B. Grad-CAM                                             -> what the CNN attends to
#   C. Wavelet / HHT / Laws spatial heat maps + agreement    -> what signal processing sees
#   D. Dimension ablation + mutual information + top-k       -> how many dimensions are needed
#
# Signal descriptors are extracted ONCE per image and reused by all four sections.
# The trained model is cached, so Grad-CAM and the heat maps do not retrain.
#
# Switch backbone with one config line. Everything else is identical between
# backbones and between datasets, so results are directly comparable.
# =====================================================================================

try:
    from google.colab import drive
    drive.mount('/content/drive')
except Exception:
    pass

# !pip install timm scipy torch torchvision scikit-learn PyWavelets scikit-image seaborn

# ================================================================
# IMPORTS
# ================================================================
import os, json, random, time, warnings
import numpy as np
import torch, torch.nn as nn, torch.nn.functional as F
from torch.optim import Adam
from torch.utils.data import DataLoader, Dataset, Subset
from torchvision import transforms, models
from torchvision.datasets import ImageFolder

import scipy, scipy.signal, scipy.stats
import pywt
from scipy.signal import hilbert
from scipy.ndimage import uniform_filter
from scipy.stats import spearmanr, ttest_rel
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import mutual_info_classif
from sklearn.metrics import (accuracy_score, precision_recall_fscore_support,
                             confusion_matrix)

import matplotlib.pyplot as plt
import matplotlib.cm as cm
import seaborn as sns

warnings.filterwarnings('ignore')

# =====================================================================================
# CONFIG  — only ROOT and BACKBONE differ between the four scripts
# =====================================================================================
ROOT     = "/content/drive/MyDrive/UABandThesis/ThesisPhD/Original_Codes_Chapter5/MetaCoinsChro/"
BACKBONE = "resnet18"        # "efficientnet_b0" | "resnet18"

DATASET = os.path.basename(ROOT.rstrip("/"))
try:
    os.chdir(ROOT)                  # results and figures are written next to the data
    print(f"Working directory: {os.getcwd()}")
except Exception as e:
    print(f"chdir failed ({e}); writing outputs to the current directory")

# Both feature modes are run, in this order:
#   balanced = fixed 80/60/60, every slot index-aligned across images
#   legacy   = concatenate raw blocks and cut at 200 (the published pipeline)
FEATURE_MODES = ["balanced", "legacy"]

# ---- what to run (applies to every mode) ----
RUN_NESTED_CV   = True     # A: paper metrics
RUN_GRADCAM     = True     # B: CNN attention panels
RUN_SIGNAL_MAPS = True     # C: signal heat maps + Grad-CAM agreement
RUN_ABLATION    = True     # D: dimension curve, MI, top-k

# ---- shared ----
SEED  = 123
IMG   = 224
BATCH = 8
FOLDS = 4
FIXED_DIM = 200                 # signal budget used by sections A-C
FREEZE_EARLY    = False         # False = full fine-tune
UNFREEZE_LAST_N = 2             # stages left trainable when FREEZE_EARLY = True

# ---- E4 interpretability checkpoint handling (added for the revised runs) ----
# The checkpoint file name, the results JSON and the figure prefix now record the
# fine-tuning regime. Before this, a partial run and a full run shared one file
# name, so the second run reloaded the first run's model instead of training its
# own, and the two regimes produced the same Grad-CAM maps.
REGIME = "partial" if FREEZE_EARLY else "full"
FORCE_RETRAIN_E4 = True         # True = ignore any existing checkpoint and retrain
# Trainable parameters of the hybrid model for two classes, as reported in the
# paper; four-class runs add 514 for the wider output layer. Verification only.
PARAM_REF = {("resnet18",        "partial"):  8575938,
             ("resnet18",        "full"):    11358722,
             ("efficientnet_b0", "partial"):   790978,
             ("efficientnet_b0", "full"):     4386366}

# ---- A: nested CV ----
GRID = [{"lr":lr, "wd":wd, "epochs":ep, "patience":3}
        for lr in (0.001, 0.002) for wd in (0.0005, 0.001) for ep in (15, 20)]

# ---- B/C: single model for interpretability ----
TRAIN_EPOCHS = 25
LR           = 1e-4
N_GRADCAM    = 8                # sherds in the Grad-CAM figure
N_MAPS       = 4                # sherds in the heat-map figure
DRAW_SAMPLED_LINES = True       # mark the 10 lines the HHT features come from

# ---- D: ablation ----
BUDGETS  = [0, 25, 50, 100, 200, 300, 400]      # 0 = CNN only (baseline)
PROPS    = (0.40, 0.30, 0.30)                   # wavelet / HHT / Laws share
ABL_HP   = {"lr": 0.001, "wd": 0.0005, "epochs": 20, "patience": 3}
TOPK_LIST = [10, 25, 50, 100]
REPEATS   = 1
SCALER_PER_FOLD = True

random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("="*70)
print(f"Dataset : {DATASET}")
print(f"Backbone: {BACKBONE}")
print(f"Regime  : {REGIME} (FREEZE_EARLY={FREEZE_EARLY}, UNFREEZE_LAST_N={UNFREEZE_LAST_N})")
print(f"Modes   : {' -> '.join(FEATURE_MODES)} | device: {device}")
print(f"Sections: " + ", ".join(n for n,f in
      [("A nested CV",RUN_NESTED_CV), ("B Grad-CAM",RUN_GRADCAM),
       ("C heat maps",RUN_SIGNAL_MAPS), ("D ablation",RUN_ABLATION)] if f))
print("="*70)

# =====================================================================================
# SIGNAL DESCRIPTORS
# Native aligned blocks: wavelet 112, HHT 200, Laws 175 (487 total).
# Every slot has a fixed meaning; missing IMFs are zero-filled rather than
# shifting later features, so truncating a block only drops its tail.
# =====================================================================================
NATIVE = {"wavelet": 112, "hht": 200, "laws": 175}
LAWS_1D = {"L5":np.array([1,4,6,4,1]), "E5":np.array([-1,-2,0,2,1]),
           "S5":np.array([-1,0,2,0,-1]), "W5":np.array([-1,2,0,-2,1]),
           "R5":np.array([1,-4,6,-4,1])}
LAWS_NAMES = [f"{a}x{b}" for a in LAWS_1D for b in LAWS_1D]
STAT7 = ["mean|c|","std","var","energy","entropy","skew","kurt"]
STAT4 = ["ampMean","ampStd","freqMean","freqStd"]

LAWS_MEANING = {
    "L5xL5":"smooth / uniform surface", "L5xE5":"linear painted elements, incisions",
    "L5xS5":"spots, pores, temper particles", "L5xW5":"wavy manufacturing traces",
    "L5xR5":"fine repetitive micro-texture",
    "E5xE5":"edge density: tool marks, motif boundaries",
    "E5xS5":"decoration disruptions", "S5xS5":"granular fabric, temper dispersion",
    "W5xR5":"firing irregularities, rippled weathering",
    "R5xR5":"high-frequency surface roughness",
}

def _entropy(x):
    p = np.abs(np.asarray(x)).flatten(); t = p.sum()
    if t <= 0: return 0.0
    p = p/t; p = p[p > 0]
    return float(-np.sum(p*np.log2(p)))

def _stats7(v):
    v = np.asarray(v).flatten()
    return [float(np.mean(np.abs(v))), float(np.std(v)), float(np.var(v)),
            float(np.mean(v**2)), _entropy(v),
            float(scipy.stats.skew(v)), float(scipy.stats.kurtosis(v))]

def _emd(sig, n=5):
    imfs=[]; res=np.asarray(sig, dtype=float).copy()
    for _ in range(n):
        if len(res) < 10: break
        pk = scipy.signal.find_peaks(res)[0]; vl = scipy.signal.find_peaks(-res)[0]
        if len(pk) < 2 or len(vl) < 2: break
        up = np.interp(range(len(res)), pk, res[pk]); lo = np.interp(range(len(res)), vl, res[vl])
        imf = res - (up+lo)/2; imfs.append(imf); res = res - imf
        if np.std(imf) < 0.01: break
    return imfs

def sampled_line_indices(gray, axis, n_lines=5):
    """The exact rows/columns the HHT features are computed from."""
    n = gray.shape[0] if axis == "row" else gray.shape[1]
    return list(range(0, n, max(1, n//10)))[:n_lines]

def wavelet_block(gray):
    """4 scales x 4 sub-bands x 7 statistics = 112"""
    f, names = [], []
    for lvl in range(1, 5):
        c = pywt.wavedec2(gray, "db4", level=lvl)
        for bname, band in zip(["LL","LH","HL","HH"], (c[0], c[1][0], c[1][1], c[1][2])):
            f += _stats7(band)
            names += [f"W-L{lvl}-{bname}-{s}" for s in STAT7]
    return np.array(f), names

def hht_block(gray, n_imfs=5, n_lines=5):
    """2 directions x 5 lines x 5 IMF slots x 4 statistics = 200 (zero-filled)"""
    f, names = [], []
    for axis in ["row","col"]:
        idx = sampled_line_indices(gray, axis, n_lines)
        for li in range(n_lines):
            if li < len(idx):
                i = idx[li]
                imfs = _emd(gray[i,:] if axis=="row" else gray[:,i], n_imfs)
            else:
                imfs = []
            for k in range(n_imfs):
                if k < len(imfs):
                    an = hilbert(imfs[k]); a = np.abs(an)
                    fr = np.diff(np.unwrap(np.angle(an)))/(2*np.pi)
                    f += [float(np.mean(a)), float(np.std(a)),
                          float(np.mean(fr)) if len(fr) else 0.0,
                          float(np.std(fr))  if len(fr) else 0.0]
                else:
                    f += [0.0, 0.0, 0.0, 0.0]
                names += [f"H-{axis}{li+1}-IMF{k+1}-{s}" for s in STAT4]
    return np.array(f), names

def laws_block(gray):
    """25 kernels x 7 statistics = 175"""
    f, names = [], []
    for name in LAWS_NAMES:
        a, b = name.split("x")
        r = scipy.signal.convolve2d(gray, np.outer(LAWS_1D[a], LAWS_1D[b]), mode="valid")
        f += _stats7(np.abs(r))
        names += [f"L-{name}-{s}" for s in STAT7]
    return np.array(f), names

def legacy_blocks(gray):
    """The descriptors the published pipeline produces (variable HHT length)."""
    c = pywt.wavedec2(gray, "db4", level=4); w=[]
    for i, coeff in enumerate(c):
        if i == 0:
            w += [np.mean(coeff), np.std(coeff), np.var(coeff),
                  scipy.stats.skew(coeff.flatten()), scipy.stats.kurtosis(coeff.flatten())]
        else:
            for d in coeff:
                w += [np.mean(d), np.std(d), np.var(d), np.max(d), np.min(d),
                      scipy.stats.skew(d.flatten()), scipy.stats.kurtosis(d.flatten())]
    h=[]
    for axis in ["row","col"]:
        for i in sampled_line_indices(gray, axis):
            for imf in _emd(gray[i,:] if axis=="row" else gray[:,i]):
                an = hilbert(imf); a = np.abs(an)
                fr = np.diff(np.unwrap(np.angle(an)))/(2*np.pi)
                h += [np.mean(a), np.std(a),
                      np.mean(fr) if len(fr) else 0, np.std(fr) if len(fr) else 0]
    l=[]
    for name in LAWS_NAMES:
        a, b = name.split("x")
        t = np.abs(scipy.signal.convolve2d(gray, np.outer(LAWS_1D[a], LAWS_1D[b]), mode="valid"))
        l += [np.mean(t), np.std(t), np.var(t), np.max(t), np.min(t),
              scipy.stats.skew(t.flatten()), scipy.stats.kurtosis(t.flatten())]
    return np.array(w), np.array(h), np.array(l)

# ---------- budget allocation ----------
def allocate(budget, props=PROPS, native=NATIVE):
    if budget <= 0: return {"wavelet":0, "hht":0, "laws":0}
    keys = ["wavelet","hht","laws"]
    want = {k:int(round(budget*p)) for k,p in zip(keys, props)}
    want[keys[-1]] = budget - want[keys[0]] - want[keys[1]]
    for _ in range(4):
        over = 0
        for k in keys:
            if want[k] > native[k]: over += want[k]-native[k]; want[k] = native[k]
        if over == 0: break
        room = {k: native[k]-want[k] for k in keys}
        tr = sum(room.values())
        if tr == 0: break
        for k in keys:
            want[k] += min(room[k], int(round(over*room[k]/tr)))
    return want

def fit_len(v, n):
    if n <= 0: return np.zeros(0)
    if len(v) > n: return v[:n]
    if len(v) < n: return np.concatenate([v, np.zeros(n-len(v))])
    return v

def build_vector(blocks, budget, mode="balanced"):
    w,h,l = blocks
    if budget <= 0: return np.zeros(0)
    if mode == "legacy":
        return fit_len(np.concatenate([w,h,l]), budget)
    a = allocate(budget)
    return np.concatenate([fit_len(w,a["wavelet"]), fit_len(h,a["hht"]), fit_len(l,a["laws"])])

def build_names(names, budget, mode="balanced"):
    wn,hn,ln = names
    if mode == "legacy": return (wn+hn+ln)[:budget]
    a = allocate(budget)
    return wn[:a["wavelet"]] + hn[:a["hht"]] + ln[:a["laws"]]

# =====================================================================================
# SPATIAL MAPS (section C)
# The 200-dim vector stores GLOBAL statistics of these maps, so the maps show the
# spatial evidence those statistics summarize, not a per-pixel contribution.
# =====================================================================================
def norm01(m):
    m = np.asarray(m, dtype=float)
    lo, hi = np.percentile(m, 1), np.percentile(m, 99)
    return np.clip((m-lo)/(hi-lo+1e-8), 0, 1)

def wavelet_level_maps(gray, levels_shown=(1,3), wavelet="db4", levels=4, win=9):
    coeffs = pywt.wavedec2(gray, wavelet, level=levels)
    out = {}
    for lvl in levels_shown:
        idx = len(coeffs)-lvl
        z = [np.zeros_like(coeffs[0])] + [tuple(np.zeros_like(d) for d in c) for c in coeffs[1:]]
        z[idx] = coeffs[idx]
        rec = pywt.waverec2(z, wavelet)[:gray.shape[0], :gray.shape[1]]
        out[f"Wavelet L{lvl}"] = uniform_filter(np.abs(rec), size=win)
    return out

def laws_energy_maps(gray, families=(("L5","S5"),("E5","E5")), win=15):
    out = {}
    for a,b in families:
        r1 = scipy.signal.convolve2d(gray, np.outer(LAWS_1D[a],LAWS_1D[b]),
                                     mode="same", boundary="symm")
        if a != b:
            r2 = scipy.signal.convolve2d(gray, np.outer(LAWS_1D[b],LAWS_1D[a]),
                                         mode="same", boundary="symm")
            resp = (np.abs(r1)+np.abs(r2))/2
        else:
            resp = np.abs(r1)
        out[f"Laws {a}x{b}"] = uniform_filter(resp, size=win)
    return out

def hht_amplitude_map(gray, imf_index=0, step=2, win=9):
    """Dense HHT map for visualization; the model uses only 5 rows + 5 columns."""
    H,W = gray.shape
    rm = np.zeros((H,W)); cmap_ = np.zeros((H,W))
    for i in range(0,H,step):
        im = _emd(gray[i,:])
        if len(im) > imf_index: rm[i,:] = np.abs(hilbert(im[imf_index]))
    for j in range(0,W,step):
        im = _emd(gray[:,j])
        if len(im) > imf_index: cmap_[:,j] = np.abs(hilbert(im[imf_index]))
    if step > 1:
        for i in range(H):
            if i % step: rm[i,:] = rm[(i//step)*step,:]
        for j in range(W):
            if j % step: cmap_[:,j] = cmap_[:,(j//step)*step]
    return uniform_filter((rm+cmap_)/2, size=win)

def all_signal_maps(gray):
    maps = {}
    maps.update(wavelet_level_maps(gray))
    maps["HHT IMF1 amplitude"] = hht_amplitude_map(gray)
    maps.update(laws_energy_maps(gray))
    maps["Signal composite"] = np.mean([norm01(v) for v in maps.values()], axis=0)
    return maps

def overlay(rgb, heat, alpha=0.45):
    H,W = rgb.shape[:2]
    h = torch.tensor(np.asarray(heat, dtype=np.float32))[None,None]
    h = F.interpolate(h, size=(H,W), mode="bilinear", align_corners=False)[0,0].numpy()
    return np.clip((1-alpha)*rgb + alpha*cm.jet(norm01(h))[:,:,:3], 0, 1)

# =====================================================================================
# DATA — descriptors extracted once, reused by every section
# =====================================================================================
MEAN = torch.tensor([0.485,0.456,0.406]).view(3,1,1)
STD  = torch.tensor([0.229,0.224,0.225]).view(3,1,1)

def denorm_hwc(x):  return torch.clamp(x*STD+MEAN, 0, 1).permute(1,2,0).numpy()

def load_everything():
    tfm = transforms.Compose([
        transforms.Resize((IMG,IMG)), transforms.ToTensor(),
        transforms.Normalize((0.485,0.456,0.406),(0.229,0.224,0.225))])
    splits = {s: ImageFolder(os.path.join(ROOT,s), transform=tfm) for s in ["train","test"]}
    classes = splits["train"].classes

    images, labels, aligned, legacy, split_of = [], [], [], [], []
    names = None
    t0 = time.time()
    total = sum(len(d) for d in splits.values())
    for sname, ds in splits.items():
        for i in range(len(ds)):
            x, y = ds[i]
            images.append(x); labels.append(y); split_of.append(sname)
            gray = np.mean(denorm_hwc(x), axis=2)
            w,wn = wavelet_block(gray); h,hn = hht_block(gray); l,ln = laws_block(gray)
            aligned.append((w,h,l))
            legacy.append(legacy_blocks(gray))
            if names is None: names = (wn,hn,ln)
            if len(images) % 10 == 0: print(f"  extracted {len(images)}/{total}")
    print(f"Descriptor extraction: {time.time()-t0:.1f}s for {total} images")
    print(f"  aligned block lengths: wavelet={len(aligned[0][0])}, "
          f"HHT={len(aligned[0][1])}, Laws={len(aligned[0][2])}")
    lw,lh,ll = legacy[0]
    print(f"  legacy raw lengths (first image): wavelet={len(lw)}, HHT={len(lh)}, Laws={len(ll)}")

    # How stable is the legacy layout across the corpus? Wavelet and Laws are fixed;
    # HHT depends on how many IMFs EMD finds, which decides where the cut at
    # FIXED_DIM falls and therefore what each slot means for each image.
    hht_lens = sorted({len(b[1]) for b in legacy})
    print(f"  legacy HHT length across images: {hht_lens[0]}–{hht_lens[-1]} "
          f"({len(hht_lens)} distinct value(s))")
    fills = []
    for b in legacy:
        w_, h_, l_ = (len(b[0]), len(b[1]), len(b[2]))
        fills.append((min(w_, FIXED_DIM),
                      max(0, min(h_, FIXED_DIM - w_)),
                      max(0, FIXED_DIM - w_ - h_)))
    wmin, wmax = min(f[0] for f in fills), max(f[0] for f in fills)
    hmin, hmax = min(f[1] for f in fills), max(f[1] for f in fills)
    lmin, lmax = min(f[2] for f in fills), max(f[2] for f in fills)
    print(f"  legacy {FIXED_DIM}-slot fill across images: wavelet {wmin}-{wmax}, "
          f"HHT {hmin}-{hmax}, Laws {lmin}-{lmax}")
    if lmax == 0:
        print("  NOTE: at this budget the cut is reached before Laws, so Laws contributes")
        print("        nothing in legacy mode.")
    if len(hht_lens) > 1:
        print("  NOTE: because the HHT length varies, a given slot index holds different")
        print("        descriptors for different sherds in legacy mode. The balanced mode")
        print("        fixes this with an 80/60/60 layout.")
    return (torch.stack(images), np.array(labels), aligned, legacy, names,
            classes, np.array(split_of))

class VecDataset(Dataset):
    def __init__(self, images, vectors, labels):
        self.images, self.vectors, self.labels = images, vectors, labels
    def __len__(self): return len(self.labels)
    def __getitem__(self, i):
        return (self.images[i], torch.tensor(self.vectors[i], dtype=torch.float32),
                int(self.labels[i]))

def collate(batch):
    xs, ss, ys = zip(*batch)
    return torch.stack(xs), torch.stack(ss), torch.tensor(ys)

# =====================================================================================
# MODEL — one class, both backbones; trunk[-1] is the Grad-CAM target
# =====================================================================================
def build_trunk(name, freeze_early=FREEZE_EARLY, unfreeze_last_n=UNFREEZE_LAST_N, verbose=False):
    name = name.lower()
    if name.startswith("efficientnet"):
        from torchvision.models import EfficientNet_B0_Weights
        m = models.efficientnet_b0(weights=EfficientNet_B0_Weights.DEFAULT)
        trunk = m.features                                   # 9 stages
    elif name == "resnet18":
        from torchvision.models import ResNet18_Weights
        m = models.resnet18(weights=ResNet18_Weights.DEFAULT)
        # conv1, bn1, relu, maxpool, layer1..layer4 — drops avgpool and fc,
        # so trunk[-1] = layer4, the counterpart of EfficientNet's last block
        trunk = nn.Sequential(*list(m.children())[:-2])
    else:
        raise ValueError(f"unsupported backbone: {name}")

    if freeze_early:
        for p in trunk.parameters(): p.requires_grad = False
        n = len(trunk)
        for i in range(max(0, n-unfreeze_last_n), n):
            for p in trunk[i].parameters(): p.requires_grad = True

    trunk.eval()
    with torch.no_grad():
        vdim = int(nn.AdaptiveAvgPool2d(1)(trunk(torch.zeros(1,3,IMG,IMG))).flatten(1).shape[1])
    if verbose:
        tr = sum(p.numel() for p in trunk.parameters() if p.requires_grad)
        tot = sum(p.numel() for p in trunk.parameters())
        print(f"  trunk '{name}': {len(trunk)} stages, vdim={vdim}, "
              f"trainable {tr:,}/{tot:,}")
    return trunk, vdim

class SignalNet(nn.Module):
    def __init__(self, num_classes, sdim, backbone=BACKBONE, verbose=False):
        super().__init__()
        self.features, vdim = build_trunk(backbone, verbose=verbose)
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.sdim = sdim
        if sdim > 0:
            self.signal = nn.Sequential(
                nn.Linear(sdim,128), nn.ReLU(), nn.Dropout(0.3),
                nn.Linear(128,64),  nn.ReLU(), nn.Dropout(0.2))
            head_in = vdim + 64
        else:
            self.signal = None; head_in = vdim
        self.head = nn.Sequential(
            nn.Linear(head_in,256), nn.ReLU(), nn.Dropout(0.4),
            nn.Linear(256,num_classes))
        self.vdim = vdim

    def forward(self, x, s):
        v = self.pool(self.features(x)).flatten(1)
        if self.signal is not None:
            v = torch.cat([v, self.signal(s)], dim=1)
        return self.head(v)

def train_model(model, loader, hp):
    crit = nn.CrossEntropyLoss()
    opt  = Adam(filter(lambda p: p.requires_grad, model.parameters()),
                lr=hp["lr"], weight_decay=hp["wd"])
    model.train(); best=float("inf"); bad=0
    for _ in range(hp["epochs"]):
        tot=0; nb=0
        for x,s,y in loader:
            x,s,y = x.to(device), s.to(device), y.to(device)
            opt.zero_grad(); loss = crit(model(x,s), y); loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0); opt.step()
            tot += loss.item(); nb += 1
        avg = tot/nb if nb else float("inf")
        if avg < best: best, bad = avg, 0
        else:
            bad += 1
            if bad >= hp.get("patience",3): break
    return model

@torch.no_grad()
def evaluate(model, loader, nC):
    model.eval(); P,T=[],[]
    for x,s,y in loader:
        P.extend(model(x.to(device), s.to(device)).argmax(1).cpu().numpy()); T.extend(y.numpy())
    acc = accuracy_score(T,P)
    pr,rc,f1,_ = precision_recall_fscore_support(T,P,average="weighted",zero_division=0)
    cmat = confusion_matrix(T,P,labels=list(range(nC)))
    return acc, pr, rc, f1, cmat

class GradCAM:
    def __init__(self, model, target_layer):
        self.model=model; self.acts=None; self.grads=None
        target_layer.register_forward_hook(lambda m,i,o: setattr(self,"acts",o.detach()))
        target_layer.register_full_backward_hook(lambda m,gi,go: setattr(self,"grads",go[0].detach()))
    def __call__(self, x, s, cls=None):
        self.model.zero_grad()
        x = x.clone().requires_grad_(True)
        out = self.model(x, s)
        if cls is None: cls = out.argmax(1)
        out.gather(1, cls.view(-1,1)).sum().backward()
        w = self.grads.mean(dim=(2,3), keepdim=True)
        cam = F.relu((w*self.acts).sum(1))
        cam = cam / (cam.amax(dim=(1,2), keepdim=True) + 1e-8)
        return cam.cpu().numpy(), cls.cpu().numpy(), out.softmax(1).detach().cpu().numpy()

# =====================================================================================
# LOAD — descriptors extracted once, reused by every mode and every section
# =====================================================================================
images, labels, aligned, legacy, names, classes, split_of = load_everything()
nC = len(classes)
n_train = int((split_of=="train").sum()); n_test = int((split_of=="test").sum())
print(f"\nClasses: {classes} | train {n_train} | test {n_test} | total {len(labels)}")
print(f"Label distribution: {dict(zip(*np.unique(labels, return_counts=True)))}")

test_idx  = np.where(split_of=="test")[0]
train_idx = np.where(split_of=="train")[0]

def run_pipeline(FEATURE_MODE):
    """Run sections A-D for one feature mode. Returns the results dict."""
    TAG          = f"{DATASET}_{BACKBONE}_{FEATURE_MODE}"
    CKPT         = os.path.join(ROOT, f"model_{TAG}_{REGIME}.pt")
    RESULTS_JSON = f"results_{TAG}_{REGIME}.json"
    FIG_PREFIX   = f"fig_{TAG}_{REGIME}"

    print("\n\n" + "#"*70)
    print(f"#  FEATURE MODE: {FEATURE_MODE.upper()}   ({TAG})")
    print("#"*70)

    BLOCKS = legacy if FEATURE_MODE == "legacy" else aligned
    V_MAIN = np.array([build_vector(b, FIXED_DIM, FEATURE_MODE) for b in BLOCKS])
    print(f"Main signal vector: {V_MAIN.shape[1]}-dim ({FEATURE_MODE} mode)")
    if FEATURE_MODE == "legacy":
        lw, lh, ll = legacy[0]
        cw = min(len(lw), FIXED_DIM)
        ch = max(0, min(len(lh), FIXED_DIM-len(lw)))
        cl = max(0, FIXED_DIM-len(lw)-len(lh))
        print(f"  first image fills the {FIXED_DIM} slots as: wavelet={cw}, HHT={ch}, Laws={cl}"
              f"  ({max(0, len(lw)+len(lh)+len(ll)-FIXED_DIM)} descriptors discarded)")
        if cl == 0:
            print("  NOTE: Laws contributes nothing at this budget in legacy mode.")
        print("  NOTE: HHT length depends on how many IMFs EMD finds, so the meaning of a")
        print("        given slot shifts between images. This is the published behaviour.")

    RESULTS = {"dataset": DATASET, "backbone": BACKBONE,
               "feature_mode": FEATURE_MODE, "signal_dim": int(V_MAIN.shape[1]),
               "seed": SEED, "folds": FOLDS, "classes": classes,
               "n_train": n_train, "n_test": n_test,
               "freeze_early": FREEZE_EARLY}

    # =====================================================================================
    # SECTION A — NESTED CROSS-VALIDATION (the paper's metrics)
    # =====================================================================================
    if RUN_NESTED_CV:
        print("\n" + "="*70)
        print(f"A. NESTED {FOLDS}-FOLD CROSS-VALIDATION — {BACKBONE} + signal")
        print("="*70)

        pool = list(range(len(labels)))
        R = {k: [] for k in ["train_acc","val_acc","test_acc","test_pr","test_rc","test_f1"]}
        R["cms"]=[]; R["best"]=[]

        for fold in range(FOLDS):
            print(f"\n{'-'*60}\nFOLD {fold+1}/{FOLDS}\n{'-'*60}")
            tr, va = train_test_split(pool, test_size=0.2, stratify=labels,
                                      random_state=SEED+fold)
            itr, iva = train_test_split(tr, test_size=0.2,
                                        stratify=[labels[i] for i in tr], random_state=SEED+fold)
            sc = StandardScaler().fit(V_MAIN[tr])
            V  = sc.transform(V_MAIN)
            ds = VecDataset(images, V, labels)
            mk = lambda idx, sh: DataLoader(Subset(ds, idx), batch_size=BATCH,
                                            shuffle=sh, collate_fn=collate)
            print(f"outer {len(tr)}/{len(va)} | inner {len(itr)}/{len(iva)}")

            best_score, best_hp = -1.0, GRID[0]
            for gi, hp in enumerate(GRID):
                m = SignalNet(nC, V.shape[1]).to(device)
                m = train_model(m, mk(itr, True), hp)
                a,_,_,_,_ = evaluate(m, mk(iva, False), nC)
                print(f"  [{gi+1}/{len(GRID)}] lr={hp['lr']} wd={hp['wd']} ep={hp['epochs']} -> {a:.4f}")
                if a > best_score: best_score, best_hp = a, hp
                del m; torch.cuda.empty_cache()
            print(f"  best: {best_hp} (inner val {best_score:.4f})")

            final = SignalNet(nC, V.shape[1], verbose=(fold==0)).to(device)
            final = train_model(final, mk(tr, True), best_hp)

            tr_a,_,_,_,tr_cm = evaluate(final, mk(tr, False), nC)
            va_a,_,_,_,va_cm = evaluate(final, mk(va, False), nC)
            te_a,te_p,te_r,te_f,te_cm = evaluate(final, mk(test_idx.tolist(), False), nC)
            print(f"  train {tr_a:.4f} | val {va_a:.4f} | TEST {te_a:.4f}")

            for k,v in [("train_acc",tr_a),("val_acc",va_a),("test_acc",te_a),
                        ("test_pr",te_p),("test_rc",te_r),("test_f1",te_f)]:
                R[k].append(v)
            R["cms"].append((tr_cm,va_cm,te_cm)); R["best"].append(best_hp)
            del final; torch.cuda.empty_cache()

        acc = np.array(R["test_acc"]); f1 = np.array(R["test_f1"])
        cv  = acc.std()/acc.mean() if acc.mean() else 0.0
        print("\n" + "="*70)
        print(f"NESTED CV SUMMARY — {BACKBONE}")
        print("="*70)
        print(f"Per-fold test accuracy: {[f'{a:.4f}' for a in acc]}")
        print(f"Test Accuracy : {acc.mean():.1%} ± {acc.std():.1%}  (CV {cv:.3f})")
        print(f"Test F1-Score : {f1.mean():.1%} ± {f1.std():.1%}")
        print(f"Test Precision: {np.mean(R['test_pr']):.1%} | Recall: {np.mean(R['test_rc']):.1%}")
        t,p = ttest_rel(R["val_acc"], R["test_acc"])
        print(f"Val vs Test t-test: t={t:.4f}, p={p:.4f} "
              f"({'significant' if p<0.05 else 'not significant'})")

        agg = np.sum([c[2] for c in R["cms"]], axis=0)
        per_class = {}
        print("\nPer-class test performance (aggregated):")
        for i,cn in enumerate(classes):
            tp = agg[i,i]; fp = agg[:,i].sum()-tp; fn = agg[i,:].sum()-tp
            pr = tp/(tp+fp) if tp+fp else 0; rc = tp/(tp+fn) if tp+fn else 0
            fc = 2*pr*rc/(pr+rc) if pr+rc else 0
            per_class[cn] = {"precision":float(pr),"recall":float(rc),"f1":float(fc)}
            print(f"  {cn}: P={pr:.3f}  R={rc:.3f}  F1={fc:.3f}")

        plt.figure(figsize=(14,4))
        plt.subplot(1,3,1)
        plt.plot(range(1,len(acc)+1), acc, "o-", lw=2, ms=8, color="tab:blue")
        plt.axhline(acc.mean(), color="red", ls="--", label=f"mean {acc.mean():.3f}")
        plt.xlabel("Fold"); plt.ylabel("Test accuracy"); plt.xticks(range(1,len(acc)+1))
        plt.title("Test accuracy"); plt.legend(); plt.grid(alpha=.3)
        plt.subplot(1,3,2)
        plt.plot(range(1,len(f1)+1), f1, "s-", lw=2, ms=8, color="tab:green")
        plt.axhline(f1.mean(), color="red", ls="--", label=f"mean {f1.mean():.3f}")
        plt.xlabel("Fold"); plt.ylabel("Test F1"); plt.xticks(range(1,len(f1)+1))
        plt.title("Test F1"); plt.legend(); plt.grid(alpha=.3)
        plt.subplot(1,3,3)
        sns.heatmap(agg, annot=True, fmt="d", cmap="Blues",
                    xticklabels=classes, yticklabels=classes)
        plt.title("Aggregated test confusion matrix"); plt.xlabel("Predicted"); plt.ylabel("True")
        plt.suptitle(f"A. Nested CV — {TAG}", fontweight="bold")
        plt.tight_layout(); plt.savefig(f"{FIG_PREFIX}_A_nestedcv.png", dpi=200, bbox_inches="tight")
        plt.show()

        RESULTS["nested_cv"] = {
            "per_fold_test_accuracy":[float(a) for a in acc],
            "per_fold_val_accuracy":[float(a) for a in R["val_acc"]],
            "per_fold_train_accuracy":[float(a) for a in R["train_acc"]],
            "per_fold_test_f1":[float(a) for a in f1],
            "test_accuracy_mean":float(acc.mean()), "test_accuracy_std":float(acc.std()),
            "test_f1_mean":float(f1.mean()), "test_precision_mean":float(np.mean(R["test_pr"])),
            "test_recall_mean":float(np.mean(R["test_rc"])), "test_cv":float(cv),
            "ttest_val_vs_test":{"t":float(t),"p":float(p)},
            "per_class":per_class, "best_hyperparams":R["best"]}

    # =====================================================================================
    # SECTION B/C — one interpretability model, shared by Grad-CAM and the heat maps
    # =====================================================================================
    model_bc = None
    if RUN_GRADCAM or RUN_SIGNAL_MAPS:
        print("\n" + "="*70)
        print("B/C. INTERPRETABILITY MODEL")
        print("="*70)
        sc_bc = StandardScaler().fit(V_MAIN[train_idx])
        V_bc  = sc_bc.transform(V_MAIN)
        ds_bc = VecDataset(images, V_bc, labels)

        model_bc = SignalNet(nC, V_bc.shape[1], verbose=True).to(device)

        # ---- regime check: the count must match the paper's parameter table ----
        n_tr = sum(p.numel() for p in model_bc.parameters() if p.requires_grad)
        exp  = PARAM_REF.get((BACKBONE, REGIME))
        if exp is not None:
            exp += 514 if nC == 4 else 0
        print(f"  regime: {REGIME} | trainable parameters: {n_tr:,}" +
              (f" | expected {exp:,} -> "
               f"{'MATCH' if n_tr == exp else 'MISMATCH'}" if exp else ""))

        if FORCE_RETRAIN_E4 and os.path.exists(CKPT):
            aside = CKPT + ".superseded"
            os.replace(CKPT, aside)
            print(f"  FORCE_RETRAIN_E4: moved aside {os.path.basename(aside)}")

        if os.path.exists(CKPT):
            model_bc.load_state_dict(torch.load(CKPT, map_location=device))
            print(f"  E4 MODEL SOURCE: loaded from {os.path.basename(CKPT)}")
            print(f"  WARNING: not trained by this run; "
                  f"its regime is whatever execution saved that file")
        else:
            print(f"  E4 MODEL SOURCE: trained by this run "
                  f"({TRAIN_EPOCHS} epochs, regime {REGIME})")
            sub,_ = train_test_split(train_idx.tolist(), test_size=0.2,
                                     stratify=labels[train_idx], random_state=SEED)
            loader = DataLoader(Subset(ds_bc, sub), batch_size=BATCH, shuffle=True, collate_fn=collate)
            opt = Adam(model_bc.parameters(), lr=LR, weight_decay=1e-3)
            crit = nn.CrossEntropyLoss(); model_bc.train()
            for ep in range(TRAIN_EPOCHS):
                tot=0
                for x,s,y in loader:
                    x,s,y = x.to(device), s.to(device), y.to(device)
                    opt.zero_grad(); loss=crit(model_bc(x,s),y); loss.backward()
                    torch.nn.utils.clip_grad_norm_(model_bc.parameters(),1.0); opt.step(); tot+=loss.item()
                if ep%5==0: print(f"  epoch {ep+1}/{TRAIN_EPOCHS} loss={tot/len(loader):.4f}")
            try:
                torch.save(model_bc.state_dict(), CKPT); print(f"  saved checkpoint: {CKPT}")
            except Exception as e:
                print(f"  checkpoint not saved: {e}")

        a,_,_,_,_ = evaluate(model_bc, DataLoader(Subset(ds_bc, test_idx.tolist()),
                             batch_size=BATCH, shuffle=False, collate_fn=collate), nC)
        print(f"  single-model test accuracy: {a:.4f}")
        RESULTS["interpretability_model_test_accuracy"] = float(a)
        cammer = GradCAM(model_bc, model_bc.features[-1])

        def cam_for(i):
            x = images[i:i+1].to(device)
            s = torch.tensor(V_bc[i:i+1], dtype=torch.float32).to(device)
            heat, pred, prob = cammer(x, s, None)
            cam224 = F.interpolate(torch.tensor(heat[None].astype(np.float32)),
                                   size=(IMG,IMG), mode="bilinear",
                                   align_corners=False)[0,0].numpy()
            return cam224, int(pred[0]), float(prob[0, pred[0]])

    # =====================================================================================
    # SECTION B — GRAD-CAM PANELS
    # =====================================================================================
    if RUN_GRADCAM and model_bc is not None:
        print("\n" + "="*70)
        print("B. GRAD-CAM")
        print("="*70)
        picks = []
        for i in test_idx:
            cam, pred, prob = cam_for(int(i))
            if pred == labels[i]:
                picks.append((int(i), cam, pred, prob))
            if len(picks) >= N_GRADCAM: break

        cols = 4; rows = max(1, (len(picks)+cols-1)//cols)
        plt.figure(figsize=(3.6*cols, 3.8*rows))
        for n,(i,cam,pred,prob) in enumerate(picks):
            ax = plt.subplot(rows, cols, n+1)
            ax.imshow(overlay(denorm_hwc(images[i]), cam)); ax.axis("off")
            ax.set_title(f"{classes[pred]} ({prob:.2f})", fontsize=10)
        task = "Chronological" if "Chro" in ROOT else "Geographical"
        note = ("expected focus: painted motifs, banding, decorative syntax"
                if task=="Chronological" else
                "attention on fabric and surface texture is theory-consistent")
        plt.suptitle(f"B. Grad-CAM — {task} classifier ({BACKBONE} + signal)\n{note}",
                     fontweight="bold", fontsize=12)
        plt.tight_layout(rect=[0,0,1,0.93])
        plt.savefig(f"{FIG_PREFIX}_B_gradcam.png", dpi=200, bbox_inches="tight")
        plt.show()
        print(f"  rendered {len(picks)} correctly classified test sherds")

    # =====================================================================================
    # SECTION C — SIGNAL HEAT MAPS + AGREEMENT WITH GRAD-CAM
    # =====================================================================================
    if RUN_SIGNAL_MAPS and model_bc is not None:
        print("\n" + "="*70)
        print("C. SIGNAL PROCESSING HEAT MAPS")
        print("="*70)
        panels, agreement = [], {}
        for i in test_idx:
            if len(panels) >= N_MAPS: break
            cam, pred, prob = cam_for(int(i))
            if pred != labels[i]: continue
            rgb = denorm_hwc(images[i]); gray = np.mean(rgb, axis=2)
            maps = all_signal_maps(gray)
            for name, m in maps.items():
                rho = spearmanr(cam.flatten(), m.flatten()).correlation
                a_ = cam >= np.percentile(cam,80); b_ = m >= np.percentile(m,80)
                iou = (a_&b_).sum()/max((a_|b_).sum(),1)
                agreement.setdefault(name, {"rho":[], "iou":[]})
                agreement[name]["rho"].append(float(rho)); agreement[name]["iou"].append(float(iou))
            panels.append({"rgb":rgb,"gray":gray,"cam":cam,"maps":maps,
                           "label":classes[pred],"prob":prob})

        keys = list(panels[0]["maps"].keys())
        cols = 2 + len(keys); rows = len(panels)
        fig, axes = plt.subplots(rows, cols, figsize=(2.6*cols, 2.9*rows))
        axes = np.atleast_2d(axes)
        for r,p in enumerate(panels):
            axes[r,0].imshow(p["rgb"]); axes[r,0].axis("off")
            axes[r,0].set_title(("Original\n" if r==0 else "")+f"{p['label']} ({p['prob']:.2f})",
                                fontsize=9)
            axes[r,1].imshow(overlay(p["rgb"], p["cam"])); axes[r,1].axis("off")
            if r==0: axes[r,1].set_title("Grad-CAM\n(CNN attention)", fontsize=9)
            for c,k in enumerate(keys, start=2):
                axes[r,c].imshow(overlay(p["rgb"], p["maps"][k])); axes[r,c].axis("off")
                if r==0:
                    sub = LAWS_MEANING.get(k.replace("Laws ",""), "")
                    axes[r,c].set_title(k if not sub else f"{k}\n({sub})", fontsize=8)
                if DRAW_SAMPLED_LINES and k.startswith("HHT"):
                    for i_ in sampled_line_indices(p["gray"], "row"):
                        axes[r,c].axhline(i_, color="white", lw=0.6, alpha=0.8)
                    for j_ in sampled_line_indices(p["gray"], "col"):
                        axes[r,c].axvline(j_, color="white", lw=0.6, alpha=0.8)
        plt.suptitle(
            f"C. Where each method looks — {TAG}\n"
            "Grad-CAM is learned attention; the other panels are the spatial energy the "
            "signal descriptors summarize.\nWhite lines mark the 5 rows + 5 columns the "
            "model's HHT features actually come from.",
            fontweight="bold", fontsize=11)
        plt.tight_layout(rect=[0,0,1,0.92])
        plt.savefig(f"{FIG_PREFIX}_C_maps.png", dpi=200, bbox_inches="tight")
        plt.show()

        ks = list(agreement.keys())
        rho = [float(np.mean(agreement[k]["rho"])) for k in ks]
        iou = [float(np.mean(agreement[k]["iou"])) for k in ks]
        o = np.argsort(rho)[::-1]
        ks=[ks[i] for i in o]; rho=[rho[i] for i in o]; iou=[iou[i] for i in o]

        fig, ax = plt.subplots(1,2, figsize=(13,4.2))
        ax[0].barh(ks[::-1], rho[::-1], color="tab:blue"); ax[0].axvline(0, color="grey", lw=1)
        ax[0].set_xlabel("Spearman ρ with Grad-CAM"); ax[0].set_title("Rank agreement")
        ax[0].grid(axis="x", alpha=.3)
        ax[1].barh(ks[::-1], iou[::-1], color="tab:orange")
        ax[1].set_xlabel("IoU of top-20% regions"); ax[1].set_title("Hot-region overlap")
        ax[1].grid(axis="x", alpha=.3)
        plt.suptitle(f"C. CNN attention vs signal energy — mean over {len(panels)} sherds",
                     fontweight="bold")
        plt.tight_layout(rect=[0,0,1,0.92])
        plt.savefig(f"{FIG_PREFIX}_C_agreement.png", dpi=200, bbox_inches="tight")
        plt.show()

        print(f"\n{'map':<28}{'Spearman ρ':>13}{'top-20% IoU':>14}")
        for k,r_,i_ in zip(ks,rho,iou): print(f"{k:<28}{r_:>13.3f}{i_:>14.3f}")
        RESULTS["signal_map_agreement"] = [{"map":k,"rho":r_,"iou":i_}
                                           for k,r_,i_ in zip(ks,rho,iou)]

    # =====================================================================================
    # SECTION D — DIMENSION ABLATION + MUTUAL INFORMATION + TOP-K
    # Always uses the aligned blocks, so feature names stay meaningful.
    # =====================================================================================
    def cv_score(vectors, tag=""):
        accs, f1s = [], []
        n = len(labels)
        for rep in range(REPEATS):
            for fold in range(FOLDS):
                tr, te = train_test_split(range(n), test_size=0.2, stratify=labels,
                                          random_state=SEED+fold+100*rep)
                if vectors is not None and vectors.shape[1] > 0:
                    sc = StandardScaler().fit(vectors[tr]) if SCALER_PER_FOLD \
                         else StandardScaler().fit(vectors)
                    V = sc.transform(vectors)
                else:
                    V = np.zeros((n,0))
                ds = VecDataset(images, V, labels)
                m = SignalNet(nC, V.shape[1]).to(device)
                m = train_model(m, DataLoader(Subset(ds,tr), batch_size=BATCH,
                                              shuffle=True, collate_fn=collate), ABL_HP)
                a,_,_,f,_ = evaluate(m, DataLoader(Subset(ds,te), batch_size=BATCH,
                                                   shuffle=False, collate_fn=collate), nC)
                accs.append(a); f1s.append(f)
                del m; torch.cuda.empty_cache()
            print(f"    {tag} rep{rep+1}: " + " ".join(f"{x:.3f}" for x in accs[-FOLDS:]))
        return np.array(accs), np.array(f1s)

    if RUN_ABLATION:
        print("\n" + "="*70)
        print(f"D. DIMENSION ABLATION ({FEATURE_MODE} blocks)")
        print("="*70)
        curve = []
        for D in BUDGETS:
            a_ = allocate(D)
            if 0 < sum(a_.values()) < D:
                print(f"\n  note: D={D} exceeds descriptor capacity ({sum(NATIVE.values())}); "
                      f"using {sum(a_.values())}")
            V = np.array([build_vector(b, D, FEATURE_MODE) for b in BLOCKS]) if D>0 else None
            lbl = "CNN only (no signal)" if D==0 else f"D={D}  [w{a_['wavelet']}/h{a_['hht']}/l{a_['laws']}]"
            print(f"\n  {lbl}")
            acc,f1 = cv_score(V, tag=f"D={D}")
            curve.append({"budget":D,"alloc":a_,"acc_mean":float(acc.mean()),
                          "acc_std":float(acc.std()),"f1_mean":float(f1.mean()),
                          "folds":[float(x) for x in acc]})
            print(f"    -> {acc.mean():.1%} ± {acc.std():.1%} | F1 {f1.mean():.1%}")

        print("\n" + "-"*66)
        print(f"{'budget':>8}{'wavelet':>9}{'HHT':>6}{'Laws':>6}{'test acc':>12}{'± std':>8}{'F1':>8}")
        for c in curve:
            a_=c["alloc"]
            print(f"{c['budget']:>8}{a_['wavelet']:>9}{a_['hht']:>6}{a_['laws']:>6}"
                  f"{c['acc_mean']:>11.1%}{c['acc_std']:>8.1%}{c['f1_mean']:>8.1%}")
        base = curve[0]["acc_mean"]
        best = max(curve[1:], key=lambda c:c["acc_mean"]) if len(curve)>1 else curve[0]
        knee = next((c for c in curve[1:] if c["acc_mean"] >= 0.98*best["acc_mean"]), best)
        print(f"\nBaseline (CNN only): {base:.1%}")
        print(f"Best budget: D={best['budget']} -> {best['acc_mean']:.1%} ({best['acc_mean']-base:+.1%} pts)")
        print(f"Smallest budget within 2% of best: D={knee['budget']} ({knee['acc_mean']:.1%})")

        # ---- mutual information + top-k: only meaningful when slots are aligned ----
        mi = None; names_ref = None; order = None; topk = []; D_ref = FIXED_DIM
        block_of = lambda n: {"W":"Wavelet","H":"HHT","L":"Laws"}.get(n.split("-")[0],"?")
        if FEATURE_MODE == "balanced":
            # ---- mutual information ----
            print("\n" + "="*70)
            print("D. MUTUAL INFORMATION (training folds only)")
            print("="*70)
            D_ref = FIXED_DIM if FIXED_DIM in BUDGETS else max(BUDGETS)
            V_ref = np.array([build_vector(b, D_ref, "balanced") for b in aligned])
            names_ref = build_names(names, D_ref, "balanced")
            tr_mi,_ = train_test_split(range(len(labels)), test_size=0.2,
                                       stratify=labels, random_state=SEED)
            mi = mutual_info_classif(StandardScaler().fit_transform(V_ref[tr_mi]),
                                     labels[tr_mi], random_state=SEED)
            order = np.argsort(mi)[::-1]
            block_of = lambda n: {"W":"Wavelet","H":"HHT","L":"Laws"}.get(n.split("-")[0],"?")

            print(f"\nTop 15 descriptors (of {D_ref}):")
            for r,i in enumerate(order[:15],1):
                print(f"  {r:>2}. {names_ref[i]:<28} MI={mi[i]:.4f}  [{block_of(names_ref[i])}]")
            print("\nBlock share among the most informative features:")
            for k in (20,50,100):
                k = min(k,len(order)); sel=[block_of(names_ref[i]) for i in order[:k]]
                present={b:sum(1 for n in names_ref if block_of(n)==b) for b in ["Wavelet","HHT","Laws"]}
                print(f"  top-{k:<4} " + "  ".join(
                    f"{b}: {sel.count(b):>3}/{present[b]:<3} ({sel.count(b)/k:.0%})"
                    for b in ["Wavelet","HHT","Laws"]))

            # ---- top-k ----
            topk = []
            for k in TOPK_LIST:
                if k > V_ref.shape[1]: continue
                print(f"\n  top-{k} features")
                acc,f1 = cv_score(V_ref[:, order[:k]], tag=f"top{k}")
                topk.append({"k":int(k),"acc_mean":float(acc.mean()),
                             "acc_std":float(acc.std()),"f1_mean":float(f1.mean())})
                print(f"    -> {acc.mean():.1%} ± {acc.std():.1%}")

        else:
            print("\n" + "="*70)
            print("D. MUTUAL INFORMATION — skipped in legacy mode")
            print("="*70)
            print("  In legacy mode the HHT block has a different length for every image, so")
            print("  column i of the feature matrix holds different descriptors for different")
            print("  sherds. Per-feature mutual information and top-k selection would therefore")
            print("  describe positions, not descriptors. The dimension curve above is still")
            print("  valid: it measures the published pipeline exactly as it behaves.")
        # ---- plots ----
        fig, ax = plt.subplots(1, 2, figsize=(13,4.4))
        xs=[c["budget"] for c in curve]; ys=[c["acc_mean"] for c in curve]; es=[c["acc_std"] for c in curve]
        ax[0].errorbar(xs, ys, yerr=es, marker="o", lw=2, ms=7, capsize=4, color="tab:blue")
        ax[0].axhline(base, color="tab:red", ls="--", label=f"CNN only: {base:.1%}")
        ax[0].axvline(FIXED_DIM, color="grey", ls=":", lw=1, label=f"published budget ({FIXED_DIM})")
        ax[0].set_xlabel("signal feature budget"); ax[0].set_ylabel("test accuracy")
        ax[0].set_title("How many signal dimensions are needed?"); ax[0].legend(); ax[0].grid(alpha=.3)
        if topk:
            kx=[c["k"] for c in topk]; ky=[c["acc_mean"] for c in topk]; ke=[c["acc_std"] for c in topk]
            ax[1].errorbar(kx, ky, yerr=ke, marker="s", lw=2, ms=7, capsize=4, color="tab:green")
            ax[1].axhline(base, color="tab:red", ls="--", label=f"CNN only: {base:.1%}")
            full = next((c["acc_mean"] for c in curve if c["budget"]==D_ref), None)
            if full: ax[1].axhline(full, color="tab:blue", ls="--", label=f"all {D_ref}: {full:.1%}")
            ax[1].set_xlabel("k most informative features"); ax[1].set_ylabel("test accuracy")
            ax[1].set_title("Selected features vs the full block"); ax[1].legend(); ax[1].grid(alpha=.3)
        plt.suptitle(f"D. Signal block ablation — {TAG}", fontweight="bold")
        plt.tight_layout(rect=[0,0,1,0.92])
        plt.savefig(f"{FIG_PREFIX}_D_ablation.png", dpi=200, bbox_inches="tight")
        plt.show()

        if mi is not None:
          plt.figure(figsize=(12,3.4))
          for b,col in [("Wavelet","tab:blue"),("HHT","tab:orange"),("Laws","tab:purple")]:
              idx=[i for i,n in enumerate(names_ref) if block_of(n)==b]
              plt.bar(idx, mi[idx], color=col, label=b, width=1.0)
          plt.xlabel(f"feature index (0–{D_ref-1})"); plt.ylabel("mutual information")
          plt.title(f"D. Where the label information sits — {TAG}")
          plt.legend(); plt.grid(axis="y", alpha=.3); plt.tight_layout()
          plt.savefig(f"{FIG_PREFIX}_D_mi.png", dpi=200, bbox_inches="tight")
          plt.show()

        RESULTS["ablation"] = {"dimension_curve":curve, "baseline_cnn_only":float(base),
                               "best_budget":best["budget"], "knee_budget":knee["budget"],
                               "topk_curve":topk, "props":list(PROPS), "native":NATIVE,
                               "hyperparameters":ABL_HP,
                               "top_features":([{"name":names_ref[i],"mi":float(mi[i]),
                                                "block":block_of(names_ref[i])} for i in order[:40]]
                                               if mi is not None else None)}

    # ---------------- export for this mode ----------------
    with open(RESULTS_JSON, "w") as fh:
        json.dump(RESULTS, fh, indent=2)
    print("\n" + "="*70)
    print(f"MODE DONE — {TAG}")
    print("="*70)
    if "nested_cv" in RESULTS:
        nc = RESULTS["nested_cv"]
        print(f"Nested CV test accuracy: {nc['test_accuracy_mean']:.1%} ± "
              f"{nc['test_accuracy_std']:.1%} | F1 {nc['test_f1_mean']:.1%}")
    if "ablation" in RESULTS:
        ab = RESULTS["ablation"]
        print(f"Ablation: baseline {ab['baseline_cnn_only']:.1%}, "
              f"best D={ab['best_budget']}, knee D={ab['knee_budget']}")
    print(f"Results: {RESULTS_JSON}")
    return RESULTS

# =====================================================================================
# RUN BOTH FEATURE MODES, THEN COMPARE
# =====================================================================================
ALL = {}
for mode in FEATURE_MODES:
    ALL[mode] = run_pipeline(mode)

print("\n\n" + "="*70)
print(f"MODE COMPARISON — {DATASET} / {BACKBONE}")
print("="*70)
print(f"{'mode':<12}{'signal dim':>12}{'test acc':>12}{'± std':>9}{'F1':>9}{'CV':>8}")
for mode, R in ALL.items():
    nc = R.get("nested_cv")
    if nc:
        print(f"{mode:<12}{R['signal_dim']:>12}{nc['test_accuracy_mean']:>11.1%}"
              f"{nc['test_accuracy_std']:>9.1%}{nc['test_f1_mean']:>9.1%}{nc['test_cv']:>8.3f}")
    else:
        print(f"{mode:<12}{R['signal_dim']:>12}{'(nested CV not run)':>38}")

if all("nested_cv" in R for R in ALL.values()) and len(ALL) == 2:
    b = ALL["balanced"]["nested_cv"]; l = ALL["legacy"]["nested_cv"]
    d = b["test_accuracy_mean"] - l["test_accuracy_mean"]
    print(f"\nbalanced - legacy = {d:+.1%} points")
    try:
        t2, p2 = ttest_rel(b["per_fold_test_accuracy"], l["per_fold_test_accuracy"])
        print(f"paired t-test over folds: t={t2:.4f}, p={p2:.4f} "
              f"({'significant' if p2 < 0.05 else 'not significant'})")
    except Exception:
        pass
    print("\nReading: a small difference means the blind truncation in the published")
    print("pipeline cost little accuracy; a large one means the aligned 80/60/60 layout")
    print("recovers information the cut was discarding.")

with open(f"summary_{DATASET}_{BACKBONE}_{REGIME}.json", "w") as fh:
    json.dump({"dataset": DATASET, "backbone": BACKBONE, "regime": REGIME,
               "freeze_early": FREEZE_EARLY, "unfreeze_last_n": UNFREEZE_LAST_N,
               "force_retrain_e4": FORCE_RETRAIN_E4,
               "modes": {m: R for m, R in ALL.items()}}, fh, indent=2)
print(f"\nCombined summary: summary_{DATASET}_{BACKBONE}_{REGIME}.json")

import sys, torchvision, sklearn, PIL, matplotlib
print(f"\nPython {sys.version.split()[0]} | torch {torch.__version__} | "
      f"torchvision {torchvision.__version__} | "
      f"GPU: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}")
```
